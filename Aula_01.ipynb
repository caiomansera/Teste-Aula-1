## Python Calculator

## 🧮 Bem-vindo(a) à Calculadora Python! 🧮

Este é um programa de calculadora simples construído em Python, que permite realizar as operações matemáticas básicas:

*   ➕ **Adição**
*   ➖ **Subtração**
*   ✖️ **Multiplicação**
*   ➗ **Divisão**

### Como usar:

1.  **Execute a célula de código da calculadora.**
2.  Você verá um menu de opções (1 a 5).
3.  **Digite o número correspondente à operação desejada** (ex: `1` para Somar).
4.  Em seguida, **insira os dois números** para a operação.
5.  O resultado será exibido.
6.  Você pode continuar realizando operações até **digitar `5` para Sair**.

---

```python
def add(x, y):
    return x + y

def subtract(x, y):
    return x - y

def multiply(x, y):
    return x * y

def divide(x, y):
    if y == 0:
        return "Error! Division by zero."
    return x / y

print("Selecione a operação.")
print("1. Somar")
print("2. Subtrair")
print("3. Multiplicar")
print("4. Dividir")
print("5. Sair")

while True:
    choice = input("Digite sua escolha(1/2/3/4/5): ")

    if choice in ('1', '2', '3', '4'):
        try:
            num1 = float(input("Digite o primeiro número: "))
            num2 = float(input("Digite o segundo número: "))
        except ValueError:
            print("Entrada inválida. Por favor, digite um número.")
            continue

        if choice == '1':
            print(num1, "+", num2, "=", add(num1, num2))

        elif choice == '2':
            print(num1, "-", num2, "=", subtract(num1, num2))

        elif choice == '3':
            print(num1, "*", num2, "=", multiply(num1, num2))

        elif choice == '4':
            print(num1, "/", num2, "=", divide(num1, num2))
    elif choice == '5':
        print("Saindo da calculadora. Adeus!")
        break
    else:
        print("Entrada inválida. Por favor, digite 1, 2, 3, 4 ou 5.")
```
