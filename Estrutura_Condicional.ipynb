```python
#EXEMPLO 1

num = int(input("Digite seu numero: "))
resto = num % 2
print(resto)
if resto == 0:
  print("É PAR")
if resto == 1:
  print("É ÍMPAR")
```

```python
#EXEMPLO 2

numero = int(input("Digite um número inteiro: "))

if numero % 2 == 0:
  print(f"o numero {numero} é par.")

else:
    print(f"o numero {numero} é impar.")
```

```python
#EXEMPLO 3

# Definição das medidas das camisetas (Comprimento, Largura em cm)
tamanhos = [
    {"nome":"PP", "comprimento": 66, "largura": 50},
    {"nome": "P", "comprimento": 70, "largura": 54},
    {"nome": "M", "comprimento": 72, "largura": 56},
    {"nome": "G", "comprimento": 74, "largura": 58},
    {"nome": "GG", "comprimento": 76, "largura": 60},
    {"nome": "XGG", "comprimento": 79, "largura": 63}
]

# Solicitar as medidas do usuário
comprimento = float(input("Digite o comprimento desejado (em cm): "))
largura = float(input("Digite a largura desejada (em cm): "))

tamanho_recomendado = None

# Percorrer os tamanhos para encontrar o menor que serve
for tamanho in tamanhos:
    if tamanho['comprimento'] >= comprimento and tamanho['largura'] >= largura:
        tamanho_recomendado = tamanho['nome']
        break  # Encontrou o menor tamanho que se encaixa, pode parar de verificar

# Exibir o resultado
if tamanho_recomendado:
    print(f"O tamanho recomendado para suas medidas é: {tamanho_recomendado}")
else:
    print("Nenhum tamanho encontrado para as medidas fornecidas. As medidas são maiores que XGG.")
```
