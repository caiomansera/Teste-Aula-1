````python
# EXERCÍCIO 1

Palavra = input("Digite a palavra cadastrada no sistema: ")
if Palavra == "caio":
  print("Sua palavra esta certa")
else:
  print("Sua palavra esta errada")
````

```pseudocode
INÍCIO
  EXIBIR "Digite a palavra cadastrada no sistema: "
  LER Palavra

  SE Palavra FOR IGUAL A "caio" ENTÃO
    EXIBIR "Sua palavra esta certa"
  SENÃO
    EXIBIR "Sua palavra esta errada"
FIM
```

````python
n = int(input("Digite a quantidade de numeros pares: "))
numeropar = 2
quantidade = 0

print(f"Os primeiros {n} pares são: ")

while n > quantidade:
  print(numeropar)
  numeropar += 2
  quantidade += 1
````
