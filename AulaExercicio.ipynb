````python
# EXERCÍCIO 1

Palavra = input("Digite a palavra cadastrada no sistema: ")
if Palavra == "caio":
  print("Sua palavra esta certa")
else:
  print("Sua palavra esta errada")
````

```pseudocode
  SENÃO

```

````python
N = int(input("Digite a quantidade de números pares que deseja exibir: "))

# Inicializa o primeiro número par positivo
numero_par_atual = 2

# Inicializa um contador para o número de pares encontrados
contador_pares_exibidos = 0

print(f"Os primeiros {N} números pares positivos são:")

# Loop até que N números pares tenham sido exibidos
while contador_pares_exibidos < N:
  print(numero_par_atual)
  numero_par_atual += 2  # Passa para o próximo número par
  contador_pares_exibidos += 1 # Incrementa a contagem de números pares exibidos
````

````python
n = int(input("Digite a quantidade de numeros pares: "))
numeropar = 2
quantidade = 0

print(f"Os primeiros {n} pares são: ")

while n > quantidade:
  print(numeropar)
  numeropar += 2
  quantidade += 1
````
